```python
def factorial (n):
    num = 1
    for i in range (1,n+1):
        num = num * i
    return num
print(factorial(5))
```

### filtering

```python
x = np.array([1,2,10,20])
print([x>=10])
print(x[x>=10])
print((x>=10) & (x<20))
print((x>=10) | (x<20))
print((x>=10) | (x==20))
```

```python
y = np.array([20,10,30,5])
x = np.where(y==10)
print(x)
x = np.where(y==10)
print(y[x])
x = np.where(y==10, 'GOOD','BAD')
print(x)
x = np.where(y>=10, 'GOOD','BAD')
print(x)
```
