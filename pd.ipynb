### Apply Function

### Define a function that returns
* if marks greater than 80 return A
* if marks greater than 60 return B
* if marks greater than 40 return C
* return fail

```python
def grade(marks):
    if marks > 80:
        return "A"
    elif marks > 60:
        return "B"
    elif marks > 40:
        return "C"
    else:
        return "Fail"
print(grade(40))


```

```python
df['grade']=df['marks'].apply(grade)
df
```

```python
df['Bonus']=df['marks'].apply(grade)
df
```

```python
df['Bonus']=df['marks'].apply(lambda x:x+5)
df
```

```python
students = pd.DataFrame({'student_id':[1,2,3,6],'name':['Malik','Ammar','Arfat','Adnan']})
marks = pd.DataFrame({'student_id':[1,2,3,4],'marks':[100,50,20,70]})
```

```python
pd.merge(
    students,
    marks,
    on='student_id',
    how='inner'
)
```

```python
pd.merge(
    students,
    marks,
    on='student_id',
    how='right'
)
```

```python
pd.merge(
    students,
    marks,
    on='student_id',
    how='left'
)
```

```python
pd.merge(
    students,
    marks,
    on='student_id',
    how='outer'
)
```

```python
s = pd.DataFrame({'id':[1,2,3,6],'name':['Malik','Ammar','Arfat','Adnan']})
m = pd.DataFrame({'id':[1,2,3,4],'marks':[100,50,20,70]})
```

```python
d=pd.merge(
    s,
    m,
    left_on='id',
    right_on='id',
    how='left'
)
d[['id','marks']]
```
